```python
from datasets import load_dataset, Audio

# ❗关键一步：关闭 audio decode
ds = ds.cast_column("context", Audio(decode=False))
```

```python
ds["test"].features
```

```python
sample = ds["test"][0]
print(sample)
```

```python
def clean_sample(sample):
    new_sample = {}
    for k, v in sample.items():
        if k == "context":
            # 只保留 path，去掉 bytes
            new_sample[k] = {
                "path": v.get("path", None)
            }
        else:
            new_sample[k] = v
    return new_sample

import json
print(json.dumps(clean_sample(sample), indent=2, ensure_ascii=False))
```
